# Data Loading and Preprocessing

```python
from pathlib import Path
import json
import os
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer

def get_preprocessed_data(project_name):
    with open(Path.cwd().parent.resolve() / 'PreProcessing Data' / f'preprocessed_report_{project_name}.json', 'r', encoding = 'utf-8') as f: 
        preprocessed_reports = json.load(f)
    with open(Path.cwd().parent.resolve() / 'PreProcessing Data' / f'preprocessed_src_{project_name}.json', 'r', encoding = 'utf-8') as f: 
        preprocessed_codes = json.load(f)
    return preprocessed_reports, preprocessed_codes

def get_tokenized_text(preprocessed_reports, preprocessed_codes):
    tokenized_reports = [report['summary']['unstemmed'] + report['description']['unstemmed'] for bug_id, report in preprocessed_reports.items()]
    tokenized_codes = [code['class_names']['unstemmed'] + code['attributes']['unstemmed'] + code['method_names']['unstemmed'] + code['variables']['unstemmed'] + code['comments']['unstemmed'] for file_path, code in preprocessed_codes.items()]
    return tokenized_reports, tokenized_codes

def get_reports_fixed_by_codes(preprocessed_reports, preprocessed_codes):
    reports_fixed_by_codes = [[] for _ in range(len(preprocessed_codes))]
    for report_index, (bug_id, report) in enumerate(preprocessed_reports.items()):
        for file_path in report['fixed_files']:
            try:
                reports_fixed_by_codes[list(preprocessed_codes.keys()).index(file_path)].append(report_index)
            except ValueError:
                continue

    for code_index, bug_ids in enumerate(reports_fixed_by_codes):
        reports_fixed_by_codes[code_index] = list(set(bug_ids))
    return reports_fixed_by_codes
    
def get_tfidf_vectors(tokenized_reports, tokenized_codes):
    vectorizer = TfidfVectorizer(analyzer = lambda x: x)
    vectorizer.fit(tokenized_reports + tokenized_codes)
    tfidf_reports = np.array(vectorizer.transform(tokenized_reports).todense())
    tfidf_codes = np.array(vectorizer.transform(tokenized_codes).todense())
    return tfidf_reports, tfidf_codes, vectorizer
```

```python
def get_labels(preprocessed_reports, preprocessed_codes):
    fixed_files_list = [file['fixed_files'] for bug_id, file in preprocessed_reports.items()]
    file_paths = list(preprocessed_codes.keys())
    Y = [[1 if file_path in fixed_files else 0 for file_path in file_paths] for fixed_files in fixed_files_list]
    for fixed_files in fixed_files_list:
        for fixed_file in fixed_files:
            if fixed_file not in file_paths:
                print("-" * 20)
                print(f"{fixed_file} not found")
                for file_name in file_paths:
                    if fixed_file.split('\\')[-1] == file_name.split('\\')[-1]:
                        print(f"Similar file found: {file_name}")
    Y = np.array(Y, dtype = np.float32)
    return Y
```

```python
glove_subset_path = Path.cwd() / "glove_subset.npz"
if not os.path.exists(glove_subset_path):
    def load_glove_embeddings(file_path):
        embeddings = {}
        with open(file_path, 'r', encoding='utf-8') as f:
            for line in f:
                values = line.split()
                word = values[0]
                try:
                    vector = np.asarray(values[1:], dtype="float32")
                except ValueError:
                    print(line)
                    exit()
                embeddings[word] = vector
        return embeddings
    
    glove_path = Path.cwd() / "glove.6B.300d.txt"
    glove = load_glove_embeddings(glove_path)

    words = np.array(list(glove.keys()))
    vectors = np.array(list(glove.values()), dtype="float32")
    np.savez_compressed(Path.cwd() / "glove_subset.npz", words=words, vectors=vectors)
    word2index = {word: index for index, word in enumerate(words)}

else:
    data = np.load(glove_subset_path, allow_pickle=True)
    words = data['words']
    vectors = data['vectors']
    word2index = {word: index for index, word in enumerate(words)}
```

```python
def get_embedding_vectors(tfidf, preprocessed_text, tokenized_text, vectorizer):
    embedding_vectors_avg = []
    embedding_vectors_words = []

    count = 0
    embedding_dim = vectors.shape[1]

    for i, text in enumerate(tokenized_text):
        word_vectors = []
        embedding_vector_by_words = np.zeros(embedding_dim)

        for word in text:
            if word in word2index:
                embedding_vector = vectors[word2index[word]]
            else:
                random_vector = np.random.normal(0, 0.1, embedding_dim)
                word_vectors.append(random_vector)
                embedding_vector_by_words += random_vector
                count += 1
                continue

            tfidf_weight = tfidf[i][vectorizer.vocabulary_.get(word, 0)]
            embedding_vector_by_words += tfidf_weight * embedding_vector
            word_vectors.append(embedding_vector)

        if len(text) > 0:
            embedding_vector_by_words /= len(text)
        else:
            print(f"Text {list(preprocessed_text.keys())[i]} is empty, set to zero after preprocessing.") 

        embedding_vectors_avg.append(embedding_vector_by_words)
        embedding_vectors_words.append(np.array(word_vectors))

    embedding_vectors_avg = np.array(embedding_vectors_avg, dtype=np.float32)

    print(f"embedding_vectors_avg shape: {embedding_vectors_avg.shape}")
    print(f"example word vectors per text: {[len(vectors) for vectors in embedding_vectors_words[:5]]}")

    return embedding_vectors_avg, embedding_vectors_words       
```

# Feature Engineering

```python
def get_similarity(tfidf_reports, tfidf_codes, embedding_vector_reports, embedding_vector_codes, reports_fixed_by_code):
    lexical_similarity = cosine_similarity(tfidf_reports, tfidf_codes)
    print("Lexical similarity: ", lexical_similarity.shape)

    semantic_similarity = cosine_similarity(embedding_vector_reports, embedding_vector_codes)
    print("Semantic similarity: ", semantic_similarity.shape)

    lexical_similarity_between_reports = cosine_similarity(tfidf_reports, tfidf_reports)
    print("Lexical similarity between reports: ", lexical_similarity_between_reports.shape)

    similar_bug_reports = np.zeros((len(tfidf_reports), len(tfidf_codes)), dtype=np.float32)
    for report_index in range(len(tfidf_reports)):
        for code_index in range(len(tfidf_codes)):
            fixed_reports = reports_fixed_by_code[code_index]
            if not fixed_reports:
                continue  
            similar_bug_reports[report_index, code_index] = np.mean(
                lexical_similarity_between_reports[report_index, fixed_reports]
            )

    similarity_similar_reports = np.array(similar_bug_reports, dtype=np.float32)
    print("Similar reports similarity: ", similarity_similar_reports.shape)

    similarities = np.stack(
        [lexical_similarity, semantic_similarity, similarity_similar_reports], axis=2
    ).astype(np.float32)
    print("Combined similarities shape: ", similarities.shape)
    return similarities
```

```python
def save_results(project_name, similarities, embedding_vector_words_reports, embedding_vector_words_codes, Y):
    embedding_words_reports_obj = np.array(embedding_vector_words_reports, dtype=object)
    embedding_words_codes_obj = np.array(embedding_vector_words_codes, dtype=object)

    np.savez_compressed(
        Path.cwd().parent.resolve() / 'PreProcessing Data' / f'prepared_data_{project_name}.npz',
        similarities=similarities,
        embedding_vector_words_reports=embedding_words_reports_obj,
        embedding_vector_words_codes=embedding_words_codes_obj,
        Y=Y
    )
```

# Data Preparation

```python
project_names = ['aspectj', 'tomcat', 'eclipse', 'swt', 'birt']
for project_name in project_names:
    print(f"Processing project: {project_name}")
    preprocessed_reports, preprocessed_codes = get_preprocessed_data(project_name)
    tokenized_reports, tokenized_codes = get_tokenized_text(preprocessed_reports, preprocessed_codes)
    reports_fixed_by_codes = get_reports_fixed_by_codes(preprocessed_reports, preprocessed_codes)
    tfidf_reports, tfidf_codes, vectorizer = get_tfidf_vectors(tokenized_reports, tokenized_codes)
    Y = get_labels(preprocessed_reports, preprocessed_codes)
    embedding_vector_avg_reports, embedding_vector_words_reports = get_embedding_vectors(tfidf_reports, preprocessed_reports, tokenized_reports, vectorizer)
    embedding_vector_avg_codes, embedding_vector_words_codes = get_embedding_vectors(tfidf_codes, preprocessed_codes, tokenized_codes, vectorizer)
    similarities = get_similarity(tfidf_reports, tfidf_codes, embedding_vector_avg_reports, embedding_vector_avg_codes, reports_fixed_by_codes)
    save_results(project_name, similarities, embedding_vector_words_reports, embedding_vector_words_codes, Y)
    print(f"Finished processing project: {project_name}\n")
```
